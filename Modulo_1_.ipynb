# Tabela Verdade


And

```python
print(10 > 2 and 4 < 6)    # V and V = v
print(10 > 2 and 4 > 6)    # V and F = V
print(10 < 2 and 4 < 6)    # F and V = F
print(10 < 2 and 4 > 6)    # F and F = F
```

OR

```python
print(10 > 2 or 4 < 6)    # V or V - V
print(10 > 2 or 4 > 6)    # V or F - V
print(10 < 2 or 4 < 6)    # F or V - V
print(10 < 2 or 4 > 6)    # F or F = F
```

Not

```python
not 10 > 2    # V - f
```

```python
not 10 < 2    # F - V
```

# Condicionais

```python
#if testar verdade

if 5 > 10:
  print('Verdade')
```

```python
#if/else testar verdade falso

num1 = 10
num2 = 5
if num1 > num2:
  print(' O num1 é o maior número')
else:
  print('o num2 é o maior número')
```

```python
#if / elif / else para varios testes

num1 = 10
num2 = 10

if num1 > num2:
  print(' O num1 é o maior número')
elif num1 < num2:
  print('o num2 é o maior número')
else:
  print('Os dois são iguais')
```

Condicionais com operadores logicos para comparações

```python
num1 = 10
num2 = 5
num3 = 2
if (num1 > num2) and (num1 > num3):
  print('O num1 é o maior número')
else:
  print('Os dois são iguais')
```

# Estrutura de Repetição
    

#For --> executar comandos enquanto percorre itens de uma coleção

```python
#for x in range (0) comandos simples enumarete, range cria objeto com sequencia numerica 0-9

for contador in range(11):

  print (contador)
```

```python
# For Range + Break, parar esse loop

for contador in range(1000):
  print (contador)
  if contador == 2:
    break
```

```python
lista = {"teste1","teste2"}
```

```python
#for Simples percorre a lista

for val in lista:
  print(val)
```

```python
#for com enumarate para saber em qual posição está

for i, val in enumerate(lista):
  print(i, val)
```

#While

Executar comandos enquanto uma condição for verdadeira

```python
#while condição comandos else comandos

val = 1
while (val < 6):
  val *=2
  print(val)
```

```python
#while com else

val = 4

while val < 6:
  val *=2
else:
  print("O valor de val é: ", val)
```
